```python
# ========================
# Financial Ratios
# ========================

df_fe["CREDIT_INCOME_RATIO"] = (
    df_fe["AMT_CREDIT"] /
    df_fe["AMT_INCOME_TOTAL"]
)

df_fe["ANNUITY_INCOME_RATIO"] = (
    df_fe["AMT_ANNUITY"] /
    df_fe["AMT_INCOME_TOTAL"]
)

df_fe["GOODS_CREDIT_RATIO"] = (
    df_fe["AMT_GOODS_PRICE"] /
    df_fe["AMT_CREDIT"]
)

df_fe["INCOME_PER_CHILD"] = (
    df_fe["AMT_INCOME_TOTAL"] /
    (1 + df_fe["CNT_CHILDREN"])
)
```

# Preprocessing Pipeline
Objectif

Construire un pipeline qui :

gère les missing values
encode les catégories
prépare les données pour XGBoost
reste reproductible

```python
# ========================
# Preprocessing Imports
# ========================

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import (
    StandardScaler,
    OrdinalEncoder
)
```

```python
# ========================
# Feature Type Detection
# ========================

numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print(f"Numerical Features: {len(numerical_features)}")
print(f"Categorical Features: {len(categorical_features)}")
```

```python
# ========================
# Memory Optimization
# ========================

for col in numerical_features:
    
    X_train[col] = X_train[col].astype("float32")
    X_valid[col] = X_valid[col].astype("float32")
    X_test[col] = X_test[col].astype("float32")

print("Numerical features converted to float32.")
```

```python
# ========================
# Numerical Transformer
# ========================

numerical_transformer = Pipeline([
    
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])
```

```python
# ========================
# Categorical Transformer
# ========================

categorical_transformer = Pipeline([
    
    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),
    
    (
        "encoder",
        OrdinalEncoder(
            handle_unknown="use_encoded_value",
            unknown_value=-1
        )
    )
])
```

```python
# ========================
# Column Transformer
# ========================

preprocessor = ColumnTransformer([
    
    (
        "num",
        numerical_transformer,
        numerical_features
    ),
    
    (
        "cat",
        categorical_transformer,
        categorical_features
    )
])
```

# XGBoost Model

```python
# ========================
# XGBoost Imports
# ========================

from xgboost import XGBClassifier
# ========================
# Compute Scale Pos Weight
# ========================

negative_class = (y_train == 0).sum()
positive_class = (y_train == 1).sum()

scale_pos_weight = (
    negative_class / positive_class
)

print(
    f"Scale Pos Weight: "
    f"{scale_pos_weight:.2f}"
)
```

```python
# ========================
# XGBoost Model
# ========================

xgb_model = XGBClassifier(
    
    # Core
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    
    # Regularization
    min_child_weight=5,
    gamma=1,
    
    # Sampling
    subsample=0.8,
    colsample_bytree=0.8,
    
    # Imbalance
    scale_pos_weight=scale_pos_weight,
    
    # Stability
    objective="binary:logistic",
    eval_metric="auc",
    
    # Performance
    n_jobs=-1,
    random_state=RANDOM_STATE
)
```

```python
# ========================
# Full XGBoost Pipeline
# ========================

xgb_pipeline = Pipeline([
    
    (
        "preprocessor",
        preprocessor
    ),
    
    (
        "model",
        xgb_model
    )
])

print("XGBoost pipeline created.")
```

# Threshold Optimization

```python
# ========================
# Threshold Optimization Imports
# ========================

from sklearn.metrics import confusion_matrix
# ========================
# Business Cost Configuration
# ========================

COST_FN = 2000   # Missed default
COST_FP = 150    # Good client rejected

print(f"False Negative Cost: {COST_FN}€")
print(f"False Positive Cost: {COST_FP}€")
```

```python
# ========================
# Threshold Evaluation Function
# ========================

def evaluate_threshold(
    y_true,
    y_proba,
    threshold,
    cost_fn,
    cost_fp
):
    
    y_pred = (
        y_proba >= threshold
    ).astype(int)
    
    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred
    ).ravel()
    
    recall = tp / (tp + fn)
    
    precision = tp / (tp + fp)
    
    total_cost = (
        fn * cost_fn
        +
        fp * cost_fp
    )
    
    return {
        "threshold": threshold,
        "recall": recall,
        "precision": precision,
        "fp": fp,
        "fn": fn,
        "cost": total_cost
    }
```

```python
# ========================
# Threshold Search
# ========================

thresholds = np.arange(
    0.10,
    0.91,
    0.05
)

results = []

for threshold in thresholds:
    
    metrics = evaluate_threshold(
        y_valid,
        y_valid_proba_xgb,
        threshold,
        COST_FN,
        COST_FP
    )
    
    results.append(metrics)

threshold_df = pd.DataFrame(results)

display(threshold_df)
```

```python
# ========================
# Best Threshold by Cost
# ========================

best_threshold = (
    threshold_df
    .sort_values("cost")
    .iloc[0]
)

display(best_threshold)
```

```python
# ========================
# Threshold Analysis Display
# ========================

print(
    "\nThreshold Analysis:\n"
)

for _, row in threshold_df.iterrows():
    
    print(
        f"Threshold {row['threshold']:.2f} │ "
        f"Recall {row['recall']:.2%} │ "
        f"Precision {row['precision']:.2%} │ "
        f"FN {int(row['fn'])} │ "
        f"FP {int(row['fp'])} │ "
        f"Cost {row['cost']:,.0f}€"
    )
```

```python
# ========================
# Threshold Tradeoff Plot
# ========================

plt.figure(figsize=(10,6))

plt.plot(
    threshold_df["threshold"],
    threshold_df["recall"],
    label="Recall"
)

plt.plot(
    threshold_df["threshold"],
    threshold_df["precision"],
    label="Precision"
)

plt.xlabel("Threshold")
plt.ylabel("Score")

plt.title("Threshold Optimization")

plt.legend()

plt.show()
```

# Probability Calobration 
Le modele classe bien les clients MAIS les probabilités ne sont pas forcément fiables.

Exemple: le modeld dit PD= 0.80 Mais en réalité seulement 55% des clients font défaut (mauvaise calibration)

```python
# ========================
# Calibration Imports
# ========================

from sklearn.calibration import (
    CalibratedClassifierCV,
    calibration_curve
)
# ========================
# Calibrated XGBoost Model
# ========================

calibrated_model = CalibratedClassifierCV(
    
    estimator=xgb_pipeline,
    
    method="isotonic",
    
    cv=3
)

print("Calibration model created.")
```

```python
# ========================
# Train Calibrated Model
# ========================

calibrated_model.fit(
    X_train,
    y_train
)

print("Calibrated model trained successfully.")
```

```python
# ========================
# Optuna Imports
# ========================

import optuna

from sklearn.metrics import roc_auc_score
# ========================
# Optuna Objective Function
# ========================

def objective(trial):
    
    params = {
        
        # Core
        "n_estimators": trial.suggest_int(
            "n_estimators",
            200,
            800
        ),
        
        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.01,
            0.15
        ),
        
        "max_depth": trial.suggest_int(
            "max_depth",
            3,
            10
        ),
        
        # Regularization
        "min_child_weight": trial.suggest_int(
            "min_child_weight",
            1,
            10
        ),
        
        "gamma": trial.suggest_float(
            "gamma",
            0,
            5
        ),
        
        # Sampling
        "subsample": trial.suggest_float(
            "subsample",
            0.6,
            1.0
        ),
        
        "colsample_bytree": trial.suggest_float(
            "colsample_bytree",
            0.6,
            1.0
        ),
        
        # Imbalance
        "scale_pos_weight": scale_pos_weight,
        
        # Stability
        "objective": "binary:logistic",
        "eval_metric": "auc",
        
        # Performance
        "random_state": RANDOM_STATE,
        "n_jobs": -1
    }
    
    model = XGBClassifier(**params)
    
    pipeline = Pipeline([
        
        (
            "preprocessor",
            preprocessor
        ),
        
        (
            "model",
            model
        )
    ])
    
    pipeline.fit(
        X_train,
        y_train
    )
    
    y_pred_proba = (
        pipeline
        .predict_proba(X_valid)[:, 1]
    )
    
    auc = roc_auc_score(
        y_valid,
        y_pred_proba
    )
    
    return auc
```
