### Freature Engineering

```python
df_fe = df.copy()

print(f"Working dataset shape: {df_fe.shape}")
```

```python
# ========================
# Financial Ratios & Safe Helpers
# ========================

# Home Credit encodes unemployed / missing employment as 365243.
# Clean it before deriving employment features.
df_fe["DAYS_EMPLOYED_ANOMALY"] = (
    df_fe["DAYS_EMPLOYED"] == 365243
).astype(int)

df_fe["DAYS_EMPLOYED"] = df_fe[
    "DAYS_EMPLOYED"
].replace(365243, np.nan)


def safe_divide(numerator, denominator):
    """Vectorized division that avoids infinite values."""
    result = numerator / denominator.replace(0, np.nan)
    return result.replace([np.inf, -np.inf], np.nan)


df_fe["CREDIT_INCOME_RATIO"] = safe_divide(
    df_fe["AMT_CREDIT"],
    df_fe["AMT_INCOME_TOTAL"]
)

df_fe["ANNUITY_INCOME_RATIO"] = safe_divide(
    df_fe["AMT_ANNUITY"],
    df_fe["AMT_INCOME_TOTAL"]
)

df_fe["GOODS_CREDIT_RATIO"] = safe_divide(
    df_fe["AMT_GOODS_PRICE"],
    df_fe["AMT_CREDIT"]
)

df_fe["ANNUITY_CREDIT_RATIO"] = safe_divide(
    df_fe["AMT_ANNUITY"],
    df_fe["AMT_CREDIT"]
)

df_fe["CREDIT_TERM"] = safe_divide(
    df_fe["AMT_CREDIT"],
    df_fe["AMT_ANNUITY"]
)

df_fe["DOWN_PAYMENT_RATIO"] = safe_divide(
    df_fe["AMT_CREDIT"] - df_fe["AMT_GOODS_PRICE"],
    df_fe["AMT_CREDIT"]
)

df_fe["INCOME_PER_CHILD"] = safe_divide(
    df_fe["AMT_INCOME_TOTAL"],
    1 + df_fe["CNT_CHILDREN"]
)

df_fe["INCOME_PER_PERSON"] = safe_divide(
    df_fe["AMT_INCOME_TOTAL"],
    df_fe["CNT_FAM_MEMBERS"]
)
```

```python
# ========================
# Age, Employment & Stability Features
# ========================

df_fe["AGE_YEARS"] = (
    -df_fe["DAYS_BIRTH"] / 365
)

df_fe["YOUNG_CLIENT"] = (
    df_fe["AGE_YEARS"] < 25
).astype(int)

df_fe["SENIOR_CLIENT"] = (
    df_fe["AGE_YEARS"] >= 60
).astype(int)

df_fe["EMPLOYMENT_YEARS"] = (
    -df_fe["DAYS_EMPLOYED"] / 365
)

df_fe["EMPLOYMENT_AGE_RATIO"] = safe_divide(
    df_fe["EMPLOYMENT_YEARS"],
    df_fe["AGE_YEARS"]
)

df_fe["EMPLOYMENT_INCOME_RATIO"] = safe_divide(
    df_fe["EMPLOYMENT_YEARS"],
    df_fe["AMT_INCOME_TOTAL"]
)

df_fe["ID_PUBLISH_YEARS"] = (
    -df_fe["DAYS_ID_PUBLISH"] / 365
)

df_fe["REGISTRATION_YEARS"] = (
    -df_fe["DAYS_REGISTRATION"] / 365
)

df_fe["PHONE_CHANGE_YEARS"] = (
    -df_fe["DAYS_LAST_PHONE_CHANGE"] / 365
)

df_fe["CAR_AGE_TO_CLIENT_AGE"] = safe_divide(
    df_fe["OWN_CAR_AGE"],
    df_fe["AGE_YEARS"]
)
```

```python
# ========================
# EXT_SOURCE Features
# ========================

ext_sources = [
    "EXT_SOURCE_1",
    "EXT_SOURCE_2",
    "EXT_SOURCE_3"
]

df_fe["EXT_SOURCE_MEAN"] = (
    df_fe[ext_sources]
    .mean(axis=1)
)

df_fe["EXT_SOURCE_STD"] = (
    df_fe[ext_sources]
    .std(axis=1)
)

df_fe["EXT_SOURCE_MIN"] = (
    df_fe[ext_sources]
    .min(axis=1)
)

df_fe["EXT_SOURCE_MAX"] = (
    df_fe[ext_sources]
    .max(axis=1)
)

df_fe["EXT_SOURCE_RANGE"] = (
    df_fe["EXT_SOURCE_MAX"] - df_fe["EXT_SOURCE_MIN"]
)

df_fe["EXT_SOURCE_MISSING_COUNT"] = (
    df_fe[ext_sources]
    .isna()
    .sum(axis=1)
)

df_fe["EXT_SOURCE_PRODUCT"] = (
    df_fe[ext_sources]
    .prod(axis=1)
)

df_fe["EXT_SOURCE_WEIGHTED"] = (
    0.2 * df_fe["EXT_SOURCE_1"]
    + 0.4 * df_fe["EXT_SOURCE_2"]
    + 0.4 * df_fe["EXT_SOURCE_3"]
)

df_fe["EXT_SOURCE_CREDIT_INTERACTION"] = (
    df_fe["EXT_SOURCE_MEAN"]
    * df_fe["CREDIT_INCOME_RATIO"]
)

df_fe["EXT_SOURCE_ANNUITY_INTERACTION"] = (
    df_fe["EXT_SOURCE_MEAN"]
    * df_fe["ANNUITY_INCOME_RATIO"]
)
```

```python
# ========================
# Contact & Social Completeness
# ========================

contact_cols = [
    "FLAG_MOBIL",
    "FLAG_EMP_PHONE",
    "FLAG_WORK_PHONE",
    "FLAG_CONT_MOBILE",
    "FLAG_PHONE",
    "FLAG_EMAIL"
]

df_fe["CONTACT_COMPLETENESS"] = (
    df_fe[contact_cols]
    .sum(axis=1)
)

social_circle_cols = [
    "OBS_30_CNT_SOCIAL_CIRCLE",
    "DEF_30_CNT_SOCIAL_CIRCLE",
    "OBS_60_CNT_SOCIAL_CIRCLE",
    "DEF_60_CNT_SOCIAL_CIRCLE"
]

df_fe["SOCIAL_CIRCLE_MISSING_COUNT"] = (
    df_fe[social_circle_cols]
    .isna()
    .sum(axis=1)
)

df_fe["DEF_30_SOCIAL_RATIO"] = safe_divide(
    df_fe["DEF_30_CNT_SOCIAL_CIRCLE"],
    df_fe["OBS_30_CNT_SOCIAL_CIRCLE"]
)

df_fe["DEF_60_SOCIAL_RATIO"] = safe_divide(
    df_fe["DEF_60_CNT_SOCIAL_CIRCLE"],
    df_fe["OBS_60_CNT_SOCIAL_CIRCLE"]
)
```

```python
# ========================
# Document, Housing & Region Features
# ========================

doc_cols = [
    col for col in df_fe.columns
    if "FLAG_DOCUMENT" in col
]

df_fe["DOCUMENT_COMPLETENESS"] = (
    df_fe[doc_cols]
    .sum(axis=1)
)

building_avg_cols = [
    col for col in df_fe.columns
    if col.endswith("_AVG")
]

df_fe["BUILDING_AVG_MISSING_COUNT"] = (
    df_fe[building_avg_cols]
    .isna()
    .sum(axis=1)
)

df_fe["BUILDING_AVG_MEAN"] = (
    df_fe[building_avg_cols]
    .mean(axis=1)
)

df_fe["REGION_RATING_MEAN"] = (
    df_fe[["REGION_RATING_CLIENT", "REGION_RATING_CLIENT_W_CITY"]]
    .mean(axis=1)
)

df_fe["REGION_RATING_DIFF"] = (
    df_fe["REGION_RATING_CLIENT_W_CITY"]
    - df_fe["REGION_RATING_CLIENT"]
)
```

```python
# ========================
# Categorical Risk Flags
# ========================

high_risk_income = [
    "Unemployed",
    "Maternity leave",
    "Student"
]

df_fe["HIGH_RISK_INCOME"] = (
    df_fe["NAME_INCOME_TYPE"]
    .isin(high_risk_income)
    .astype(int)
)

stable_income = [
    "Working",
    "Commercial associate",
    "State servant",
    "Pensioner"
]

df_fe["STABLE_INCOME_TYPE"] = (
    df_fe["NAME_INCOME_TYPE"]
    .isin(stable_income)
    .astype(int)
)

low_education = [
    "Lower secondary",
    "Secondary / secondary special"
]

df_fe["LOW_EDUCATION"] = (
    df_fe["NAME_EDUCATION_TYPE"]
    .isin(low_education)
    .astype(int)
)

df_fe["HAS_CAR_AND_REALTY"] = (
    (df_fe["FLAG_OWN_CAR"] == "Y")
    & (df_fe["FLAG_OWN_REALTY"] == "Y")
).astype(int)
```

```python
# ========================
# Missingness & Data Quality Signals
# ========================

high_signal_missing_cols = [
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "OWN_CAR_AGE",
    "DAYS_LAST_PHONE_CHANGE",
    "EXT_SOURCE_1",
    "EXT_SOURCE_2",
    "EXT_SOURCE_3"
]

df_fe["HIGH_SIGNAL_MISSING_COUNT"] = (
    df_fe[high_signal_missing_cols]
    .isna()
    .sum(axis=1)
)

df_fe["TOTAL_MISSING_COUNT"] = (
    df_fe
    .isna()
    .sum(axis=1)
)

numeric_cols = df_fe.select_dtypes(
    include=["int64", "float64", "float32"]
).columns

df_fe[numeric_cols] = df_fe[numeric_cols].replace(
    [np.inf, -np.inf],
    np.nan
)

print("Advanced engineered features created and infinite values cleaned.")
```

```python
# ========================
# Remove Redundant Features
# ========================

high_corr_features = [
    "REGION_RATING_CLIENT"
]

df_fe = df_fe.drop(
    columns=high_corr_features,
    errors="ignore"
)

print("Highly correlated features removed.")
```

# Preprocessing Pipeline
Objectif

Construire un pipeline qui :

gère les missing values
encode les catégories
prépare les données pour XGBoost
reste reproductible

```python
# ========================
# Preprocessing Imports
# ========================

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import (
    StandardScaler,
    OrdinalEncoder
)
```

```python
# ========================
# Feature Type Detection
# ========================

numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print(f"Numerical Features: {len(numerical_features)}")
print(f"Categorical Features: {len(categorical_features)}")
```

```python
# ========================
# Memory Optimization
# ========================

for col in numerical_features:
    
    X_train[col] = X_train[col].astype("float32")
    X_valid[col] = X_valid[col].astype("float32")
    X_test[col] = X_test[col].astype("float32")

print("Numerical features converted to float32.")
```

```python
# ========================
# Numerical Transformer
# ========================

numerical_transformer = Pipeline([
    
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])
```

```python
# ========================
# Categorical Transformer
# ========================

categorical_transformer = Pipeline([
    
    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),
    
    (
        "encoder",
        OrdinalEncoder(
            handle_unknown="use_encoded_value",
            unknown_value=-1
        )
    )
])
```

```python
# ========================
# Column Transformer
# ========================

preprocessor = ColumnTransformer([
    
    (
        "num",
        numerical_transformer,
        numerical_features
    ),
    
    (
        "cat",
        categorical_transformer,
        categorical_features
    )
])
```

# Baseline Model — Logistic Regression

```python
# ========================
# Modeling Imports
# ========================

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)


def ks_statistic(y_true, y_proba):
    scores = pd.DataFrame({
        "target": y_true,
        "proba": y_proba
    }).sort_values("proba", ascending=False)
    positives = scores["target"].sum()
    negatives = len(scores) - positives
    scores["cum_positive_rate"] = scores["target"].cumsum() / positives
    scores["cum_negative_rate"] = (1 - scores["target"]).cumsum() / negatives
    return (
        scores["cum_positive_rate"]
        - scores["cum_negative_rate"]
    ).abs().max()
```

```python
# ========================
# Baseline Pipeline
# ========================

baseline_model = Pipeline([
    
    (
        "preprocessor",
        preprocessor
    ),
    
    (
        "model",
        LogisticRegression(
            class_weight="balanced",
            max_iter=1000,
            random_state=RANDOM_STATE
        )
    )
])

print("Baseline pipeline created.")
```

```python
# ========================
# Train Baseline Model
# ========================

baseline_model.fit(
    X_train,
    y_train
)

print("Baseline model trained successfully.")
```

```python
# ========================
# Validation Predictions
# ========================

y_valid_pred = baseline_model.predict(X_valid)

y_valid_proba = baseline_model.predict_proba(
    X_valid
)[:, 1]
```

```python
# ========================
# ROC-AUC Score
# ========================

baseline_auc = roc_auc_score(
    y_valid,
    y_valid_proba
)

print(f"Baseline ROC-AUC: {baseline_auc:.4f}")
```

```python
# ========================
# Classification Metrics
# ========================

print(
    classification_report(
        y_valid,
        y_valid_pred
    )
)
```

```python
# ========================
# Confusion Matrix
# ========================

cm = confusion_matrix(
    y_valid,
    y_valid_pred
)

print(cm)
```

```python
# ========================
# Business Metrics
# ========================

recall = recall_score(
    y_valid,
    y_valid_pred
)

precision = precision_score(
    y_valid,
    y_valid_pred
)

f1 = f1_score(
    y_valid,
    y_valid_pred
)

gini = 2 * baseline_auc - 1

print(f"Recall:   {recall:.4f}")
print(f"Precision:{precision:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"Gini:     {gini:.4f}")
```

# XGBoost Model

```python
# ========================
# XGBoost Imports
# ========================

from xgboost import XGBClassifier
# ========================
# Compute Scale Pos Weight
# ========================

negative_class = (y_train == 0).sum()
positive_class = (y_train == 1).sum()

scale_pos_weight = (
    negative_class / positive_class
)

print(
    f"Scale Pos Weight: "
    f"{scale_pos_weight:.2f}"
)
```

```python
# ========================
# XGBoost Model
# ========================

xgb_model = XGBClassifier(
    # Core
    n_estimators=650,
    learning_rate=0.035,
    max_depth=4,
    max_leaves=32,
    grow_policy="lossguide",

    # Regularization
    min_child_weight=20,
    gamma=0.5,
    reg_alpha=0.05,
    reg_lambda=5.0,
    max_delta_step=1,

    # Sampling
    subsample=0.85,
    colsample_bytree=0.85,
    colsample_bynode=0.80,

    # Imbalance
    scale_pos_weight=scale_pos_weight,

    # Stability
    objective="binary:logistic",
    eval_metric="auc",
    tree_method="hist",

    # Performance
    n_jobs=-1,
    random_state=RANDOM_STATE
)
```

```python
# ========================
# Full XGBoost Pipeline
# ========================

xgb_pipeline = Pipeline([
    
    (
        "preprocessor",
        preprocessor
    ),
    
    (
        "model",
        xgb_model
    )
])

print("XGBoost pipeline created.")
```

```python
# ========================
# Train XGBoost Model
# ========================

xgb_pipeline.fit(
    X_train,
    y_train
)

print("XGBoost model trained successfully.")
```

```python
# ========================
# Validation Predictions
# ========================

y_valid_pred_xgb = xgb_pipeline.predict(
    X_valid
)

y_valid_proba_xgb = (
    xgb_pipeline
    .predict_proba(X_valid)[:, 1]
)
```

```python
# ========================
# ROC-AUC Score
# ========================

xgb_auc = roc_auc_score(
    y_valid,
    y_valid_proba_xgb
)

print(f"XGBoost ROC-AUC: {xgb_auc:.4f}")
```

```python
# ========================
# Classification Report
# ========================

print(
    classification_report(
        y_valid,
        y_valid_pred_xgb
    )
)
```

# Threshold Optimization

```python
# ========================
# Threshold Optimization Imports
# ========================

from sklearn.metrics import confusion_matrix
# ========================
# Business Cost Configuration
# ========================

COST_FN = 2000   # Missed default
COST_FP = 150    # Good client rejected

print(f"False Negative Cost: {COST_FN}€")
print(f"False Positive Cost: {COST_FP}€")
```

```python
# ========================
# Threshold Evaluation Function
# ========================

def evaluate_threshold(
    y_true,
    y_proba,
    threshold,
    cost_fn,
    cost_fp
):
    
    y_pred = (
        y_proba >= threshold
    ).astype(int)
    
    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred
    ).ravel()
    
    recall = tp / (tp + fn)
    
    precision = tp / (tp + fp)
    
    total_cost = (
        fn * cost_fn
        +
        fp * cost_fp
    )
    
    return {
        "threshold": threshold,
        "recall": recall,
        "precision": precision,
        "fp": fp,
        "fn": fn,
        "cost": total_cost
    }
```

```python
# ========================
# Threshold Search
# ========================

thresholds = np.arange(
    0.10,
    0.91,
    0.05
)

results = []

for threshold in thresholds:
    
    metrics = evaluate_threshold(
        y_valid,
        y_valid_proba_xgb,
        threshold,
        COST_FN,
        COST_FP
    )
    
    results.append(metrics)

threshold_df = pd.DataFrame(results)

display(threshold_df)
```

```python
# ========================
# Best Threshold by Cost
# ========================

best_threshold = (
    threshold_df
    .sort_values("cost")
    .iloc[0]
)

display(best_threshold)
```

```python
# ========================
# Threshold Analysis Display
# ========================

print(
    "\nThreshold Analysis:\n"
)

for _, row in threshold_df.iterrows():
    
    print(
        f"Threshold {row['threshold']:.2f} │ "
        f"Recall {row['recall']:.2%} │ "
        f"Precision {row['precision']:.2%} │ "
        f"FN {int(row['fn'])} │ "
        f"FP {int(row['fp'])} │ "
        f"Cost {row['cost']:,.0f}€"
    )
```

```python
# ========================
# Threshold Tradeoff Plot
# ========================

plt.figure(figsize=(10,6))

plt.plot(
    threshold_df["threshold"],
    threshold_df["recall"],
    label="Recall"
)

plt.plot(
    threshold_df["threshold"],
    threshold_df["precision"],
    label="Precision"
)

plt.xlabel("Threshold")
plt.ylabel("Score")

plt.title("Threshold Optimization")

plt.legend()

plt.show()
```

# Ranking Metrics and Threshold Selection

ROC-AUC and Gini are kept as the main ranking metrics because they evaluate how well the model separates risky borrowers from safer borrowers independently of any fixed decision threshold.

Precision, recall, F1-score, false positives, false negatives, and business cost depend on the selected threshold. Therefore, threshold selection is treated as an operational decision after the ranking model is trained.


```python
# ========================
# Fine Threshold Search: F1 and Business Cost
# ========================

threshold_grid = np.arange(
    0.01,
    0.91,
    0.01
)

threshold_results = []

for threshold in threshold_grid:
    
    y_pred_threshold = (
        y_valid_proba_opt >= threshold
    ).astype(int)
    
    tn, fp, fn, tp = confusion_matrix(
        y_valid,
        y_pred_threshold
    ).ravel()
    
    threshold_results.append({
        "threshold": threshold,
        "precision": precision_score(
            y_valid,
            y_pred_threshold,
            zero_division=0
        ),
        "recall": recall_score(
            y_valid,
            y_pred_threshold
        ),
        "f1": f1_score(
            y_valid,
            y_pred_threshold,
            zero_division=0
        ),
        "fp": fp,
        "fn": fn,
        "cost": fn * COST_FN + fp * COST_FP
    })

threshold_selection_df = pd.DataFrame(
    threshold_results
)

best_f1_threshold = (
    threshold_selection_df
    .sort_values(
        "f1",
        ascending=False
    )
    .iloc[0]
)

best_cost_threshold = (
    threshold_selection_df
    .sort_values("cost")
    .iloc[0]
)

print("Best threshold by F1:")
display(best_f1_threshold)

print("Best threshold by business cost:")
display(best_cost_threshold)

display(
    threshold_selection_df.sort_values(
        "f1",
        ascending=False
    ).head(10)
)
```

## Threshold Selection Interpretation

The same XGBoost ranking model can support different business objectives depending on the selected threshold:

- Maximize ROC-AUC/Gini: evaluate ranking quality.
- Maximize F1-score: balance precision and recall for classification quality.
- Minimize business cost: choose the threshold that best reflects the cost of missed defaults versus rejected good clients.

For credit scoring, the recommended workflow is to report ROC-AUC/Gini for model quality, then select the operational threshold using either F1-score or business cost depending on the decision policy.

```python
# ========================
# Risk Bands
# ========================

def assign_risk_band(pd_value):
    
    if pd_value < 0.03:
        return "A - Low Risk"
    
    elif pd_value < 0.07:
        return "B - Moderate Risk"
    
    elif pd_value < 0.15:
        return "C - High Risk"
    
    else:
        return "D - Critical Risk"

risk_df["risk_band"] = (
    risk_df["predicted_pd"]
    .apply(assign_risk_band)
)

risk_df.head()
```

```python
# ========================
# Risk Band Distribution
# ========================

risk_distribution = (
    risk_df["risk_band"]
    .value_counts(normalize=True)
    * 100
)

print(risk_distribution)
```

# Lending Decision Interpretation

Credit decisions are based on calibrated risk bands rather than individual outcomes.

Important:
- the model estimates probabilities of default (PD)
- predictions are statistical, not deterministic
- some high-risk borrowers may still repay successfully
- some approved borrowers may still default

The objective of the scoring system is to optimize overall portfolio risk rather than perfectly predict every individual borrower outcome.


```python
# ========================
# Updated Lending Decisions
# ========================

def lending_decision(risk_band):
    
    if risk_band == "A - Low Risk":
        return "Approve"
    
    elif risk_band == "B - Moderate Risk":
        return "Approve"
    
    elif risk_band == "C - High Risk":
        return "Manual Review"
    
    else:
        return "Reject"

# Apply decision logic
risk_df["decision"] = (
    risk_df["risk_band"]
    .apply(lending_decision)
)

# Display sample results
risk_df.head()
```

```python
# =====================================================
# IMPORTS
# =====================================================
import sys
from pathlib import Path

# racine projet
ROOT_DIR = Path.cwd().parents[1]

# ajouter au PYTHONPATH
sys.path.append(str(ROOT_DIR))
import joblib
import pandas as pd

from sklearn.compose import (
    ColumnTransformer
)

from sklearn.pipeline import (
    Pipeline
)

from sklearn.preprocessing import (
    OneHotEncoder
)

from sklearn.metrics import (
    roc_auc_score
)

from xgboost import XGBClassifier


# =====================================================
# IMPORT CUSTOM TRANSFORMER
# =====================================================

from backend.ml.scoring.transformers import (
    FeatureEngineeringTransformer
)


# =====================================================
# PRODUCTION FEATURES
# Ce que le banquier peut réellement saisir
# =====================================================

production_features = [

    "AMT_INCOME_TOTAL",

    "AMT_CREDIT",

    "AMT_ANNUITY",

    "AMT_GOODS_PRICE",

    "DAYS_BIRTH",

    "DAYS_EMPLOYED",

    "NAME_EDUCATION_TYPE",

    "NAME_INCOME_TYPE",

    "CODE_GENDER",

    "CNT_CHILDREN",

    "REGION_RATING_CLIENT_W_CITY",

    "EXT_SOURCE_1",

    "EXT_SOURCE_2",

    "EXT_SOURCE_3"
]


# =====================================================
# DATASETS
# =====================================================

X_train_prod = X_train[
    production_features
]

X_test_prod = X_test[
    production_features
]


# =====================================================
# FEATURE ENGINEERING PREVIEW
# =====================================================

fe_transformer = (
    FeatureEngineeringTransformer()
)

X_train_fe = fe_transformer.fit_transform(
    X_train_prod
)


# =====================================================
# DETECT CATEGORICAL COLUMNS
# =====================================================

categorical_cols = (

    X_train_fe

    .select_dtypes(
        include=["object", "category"]
    )

    .columns

    .tolist()
)

print("Categorical columns:")
print(categorical_cols)


# =====================================================
# PREPROCESSOR
# =====================================================

preprocessor = ColumnTransformer(

    transformers=[

        (
            "cat",

            OneHotEncoder(
                handle_unknown="ignore"
            ),

            categorical_cols
        )
    ],

    remainder="passthrough"
)


# =====================================================
# XGBOOST MODEL
# =====================================================

xgb_model = XGBClassifier(

    random_state=42,

    n_estimators=300,

    max_depth=6,

    learning_rate=0.05,

    subsample=0.8,

    colsample_bytree=0.8,

    eval_metric="logloss"
)


# =====================================================
# COMPLETE PIPELINE
# =====================================================

pipeline = Pipeline([

    (
        "feature_engineering",

        FeatureEngineeringTransformer()
    ),

    (
        "preprocessor",

        preprocessor
    ),

    (
        "model",

        xgb_model
    )
])


# =====================================================
# TRAINING
# =====================================================

pipeline.fit(

    X_train_prod,

    y_train
)

print("Pipeline trained.")


# =====================================================
# PREDICTIONS
# =====================================================

y_pred_proba = (
    pipeline.predict_proba(
        X_test_prod
    )[:, 1]
)


# =====================================================
# METRICS
# =====================================================

auc = roc_auc_score(

    y_test,

    y_pred_proba
)

print(f"AUC : {auc:.4f}")


# =====================================================
# SAVE PIPELINE
# =====================================================

joblib.dump(

    pipeline,

    "../../backend/ml/scoring/scoring_pipeline.pkl"
)

joblib.dump(

    production_features,

    "../../backend/ml/scoring/features.pkl"
)

print("Pipeline saved.")
```
